# 224 — Finding a domain and drawing its edges are different problems, and the truth set runs out first

**Question:** notebooks [220](./220_mhc_region_map_and_coverage.ipynb) to
[223](./223_mhc_platform_vs_ig_domain.ipynb) all report one number per arm per species: how
many MHC domains it recovered, where "recovered" means the transferred family matched and the
interval reached IoU ≥ 0.5 against the true domain (*boundary matched*). That number fuses two
things a reader will assume are one. Did the search reach the protein at all, and did it draw
the right box around the domain? This notebook separates them, then asks how far out either
question still has an answer.

**Tools and run.** Data are the midi-plus run (998 queries, nine scored targets; see 220).
The nine comparison tools (phmmer, jackhmmer, MMseqs2, MMseqs2 iterative, HHblits, Foldseek,
ProstT5, Reseek, Folddisco) are set against seven kmerseek arms: the four HP arms
hp_pbotc_1st_ed2 at k=19, 21 and 24 and hp_thomas_dill2 at k=26, plus dayhoff6 k=12,
funcgroups8 k=12 and protein20 k=10. Where a figure pools tools it says so; where it draws one
kmerseek arm it names the alphabet and k. An *arm* is one tool at one setting.

Three claims are tested here and one of them is a limitation rather than a result.

1. Where a tool can score a true positive on an MHC molecule is decided by the Pfam
   annotation, not by divergence. The tool cliff and the annotation cliff are the same line, so
   the benchmark cannot tell "the molecules stopped being findable" apart from "Pfam stopped
   annotating them". TAP1 and TAP2 are the control, and fly, further out than ciona and yet far
   better scored on the strength of one annotated protein, is the proof.
2. Called and boundary matched come apart, and they come apart in opposite directions for
   different tools. Reporting only the favourable half of that split would be misleading in
   either direction.
3. Among kmerseek's own alphabets the HP ordering holds on a metric where kmerseek loses
   overall, which is the point: the ordering is not an artifact of a self-flattering metric.

Every denominator here is computed from the run's own `truth/*_domain_map.parquet` files. The
pipeline's `recall_reachable` and `n_reachable_instances` columns in `all_domain_metrics.parquet`
are never read, in this notebook or in 223.

Siblings: [220](./220_mhc_region_map_and_coverage.ipynb) maps the region,
[221](./221_mhc_cross_species_detection_depth.ipynb) walks it out to 2,000 Mya,
[222](./222_mhc_synteny_target_side.ipynb) asks what the hits landed on, and
[223](./223_mhc_platform_vs_ig_domain.ipynb) explains why the IoU cut behaves the way it does.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import polars as pl

import mhc_region_utils as mu
import ortholog_analysis_utils as ou

FIG_DIR = Path("../figures")
plt.rcParams.update({"figure.dpi": 120, "savefig.bbox": "tight", "font.size": 10})

gene_map = mu.load_query_gene_map()
window = mu.load_mhc_window(gene_map)
CORE = mu.core_truth()
CORE_ACC = CORE["query_acc"].unique().to_list()
WIN_ACC = window["accession"].to_list()

ARMS = mu.focus_arms()
COMPARISON_TOOLS = [t for t in ARMS["tool"].unique().to_list() if t != "kmerseek"]
COMPARISON_TEXT = ", ".join(mu.TOOL_LABELS[t] for t in mu.COMPARISON_TOOLS if t in COMPARISON_TOOLS)
STATE_LABEL = {
    "boundary_matched": "boundary matched (IoU ≥ 0.5)",
    "called only": "called, not boundary matched",
    "not called": "not called",
}
HP_ARMS = [
    "hp_pbotc_1st_ed2_k19_lcTrue",
    "hp_pbotc_1st_ed2_k21_lcTrue",
    "hp_pbotc_1st_ed2_k24_lcTrue",
    "hp_thomas_dill2_k26_lcTrue",
]
KM_ARMS = HP_ARMS + [
    "dayhoff6_k12_lcTrue",
    "funcgroups8_k12_lcTrue",
    "protein20_k10_lcTrue",
]
SP = mu.SPECIES_ORDER

print(f"MHC-window genes: {window.height}")
print(mu.RUN_SUMMARY)
print(f"curated MHC molecules + peptide-loading control: {len(CORE_ACC)} (24 chr6 genes + B2M)")
print(f"true (gene, family) pairs on those: {CORE.height}")
print(f"focus arms in the raw call table: {ARMS.height}")
print(f"species: {SP}")
print()
comp = (
    CORE.group_by("mhc_class")
    .agg(n_genes=pl.col("query_acc").n_unique(), n_domains=pl.len())
    .sort("n_domains", descending=True)
)
with pl.Config(tbl_rows=20, tbl_width_chars=140):
    print(comp)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.2))
axes[0].barh(range(comp.height), comp["n_genes"], color="#4878CF", alpha=0.9)
axes[0].set_yticks(range(comp.height))
axes[0].set_yticklabels(comp["mhc_class"], fontsize=8.5)
axes[0].set_xlabel("genes")
axes[0].set_title(f"the {len(CORE_ACC)} curated genes this notebook tracks")
sub = window.group_by("mhc_subregion").len().sort("len")
axes[1].barh(
    range(sub.height),
    sub["len"],
    color=[mu.MHC_SUBREGION_COLORS[s] for s in sub["mhc_subregion"]],
    alpha=0.9,
)
axes[1].set_yticks(range(sub.height))
axes[1].set_yticklabels(sub["mhc_subregion"], fontsize=8.5)
axes[1].set_xlabel("genes")
axes[1].set_title(f"the {window.height}-gene MHC window behind the per-tool splits")
for ax in axes:
    for s_ in ("right", "top"):
        ax.spines[s_].set_visible(False)
mu.finish_figure(
    fig,
    FIG_DIR / "224_scope.png",
    tools=mu.NO_TOOL,
    hypothesis="The curated MHC genes and the MHC window are large enough to split per tool and per species.",
    conclusion=(
        f"{len(CORE_ACC)} curated genes carrying {CORE.height} true (gene, family) pairs, inside a "
        f"{window.height}-gene window spread over five sub-regions; {ARMS.height} arms in the raw call table."
    ),
)
```

## 1. The tool cliff and the annotation cliff are the same line

Each cell below is one MHC molecule against one target proteome, in three states: the arms made
no call on that gene, they made calls but none boundary matched any of its domains, or they
boundary matched at least one. Two panels: the nine comparison tools pooled, and the four HP
kmerseek arms pooled (hp_pbotc_1st_ed2 k=19, 21, 24 and hp_thomas_dill2 k=26).

Behind each cell is the thing the state has to be read against: whether any of that gene's
Pfam families is annotated on any protein in that target proteome. A cell with no reachable
family cannot produce a true positive for any tool at any sensitivity, because there is no
instance in the answer key to match.

TAP1 and TAP2 are in the figure as the control. They sit inside the class II region but are ABC
transporters, a fold shared with bacteria, and their two Pfam families are annotated in all nine
proteomes. If detection simply died with distance they would die too.


```python
# Zero-filled grid. A (gene, family, arm, species) cell with no call has no row in the call
# table, and dropping it would shrink the denominator to the cells that already succeeded.
G = mu.domain_grid(CORE, SP)
G = G.with_columns(
    group=pl.when(pl.col("tool") == "kmerseek")
    .then(pl.lit("kmerseek HP"))
    .otherwise(pl.lit("comparison tools"))
)
G = G.filter((pl.col("group") == "comparison tools") | pl.col("variant").is_in(HP_ARMS))

# Did the arm group call anything at all on that gene, regardless of family? That is a
# different question from "did it call the right family", so it is counted separately.
any_call = (
    mu.focus_calls()
    .filter(pl.col("query_acc").is_in(CORE_ACC))
    .with_columns(
        group=pl.when(pl.col("tool") == "kmerseek")
        .then(pl.lit("kmerseek HP"))
        .otherwise(pl.lit("comparison tools"))
    )
    .filter((pl.col("group") == "comparison tools") | pl.col("variant").is_in(HP_ARMS))
    .group_by("group", "species", "query_acc")
    .agg(n_any=pl.len())
    .collect()
)

gene_state = (
    G.group_by("group", "species", "query_acc", "hgnc_symbol", "mhc_class")
    .agg(n_boundary_matched=(pl.col("best_iou") >= 0.5).sum(), n_domains=pl.len())
    .join(any_call, on=["group", "species", "query_acc"], how="left")
    .with_columns(n_any=pl.col("n_any").fill_null(0))
    .with_columns(
        state=pl.when(pl.col("n_boundary_matched") > 0)
        .then(pl.lit("boundary_matched"))
        .when(pl.col("n_any") > 0)
        .then(pl.lit("called only"))
        .otherwise(pl.lit("not called"))
    )
)

# Reachability: is any of the gene's true families annotated in that target proteome?
fams = CORE["pfam_id"].unique().to_list()
reach_fam = mu.target_family_counts(fams, SP)
reach_gene = (
    CORE.select("query_acc", "hgnc_symbol", "pfam_id")
    .join(reach_fam, on="pfam_id", how="inner")
    .group_by("query_acc", "hgnc_symbol", "species")
    .agg(
        n_reachable_families=(pl.col("n_target_proteins") > 0).sum(),
        n_families=pl.len(),
    )
)

print(
    "Pfam instances in each target proteome, per MHC family (the recall denominator):"
)
with pl.Config(tbl_rows=20, tbl_width_chars=200):
    print(
        reach_fam.pivot(on="species", index="pfam_id", values="n_target_proteins")
        .select(["pfam_id"] + SP)
        .with_columns(
            pfam=pl.col("pfam_id").replace_strict(
                mu.pfam_label(fams), default=pl.col("pfam_id")
            )
        )
    )

past = (
    G.filter(
        (pl.col("group") == "comparison tools")
        & (pl.col("best_iou") >= 0.5)
        & pl.col("species").is_in(mu.PRE_MHC_SPECIES)
    )
    .group_by("species", "pfam_id")
    .agg(
        genes_boundary_matched=pl.col("query_acc").n_unique(), tools=pl.col("tool").n_unique()
    )
    .sort("species", "pfam_id")
)
print("\nPast zebrafish, which family still carries a boundary-matched MHC-gene domain:")
with pl.Config(tbl_rows=20, tbl_width_chars=160):
    print(past)

summary = (
    gene_state.join(reach_gene, on=["query_acc", "hgnc_symbol", "species"], how="left")
    .group_by("group", "species")
    .agg(
        boundary_matched=(pl.col("state") == "boundary_matched").sum(),
        called_only=(pl.col("state") == "called only").sum(),
        not_called=(pl.col("state") == "not called").sum(),
        genes_with_a_reachable_family=(pl.col("n_reachable_families") > 0).sum(),
    )
    .sort("group", "species")
)
print(f"\nPer species, over the {len(CORE_ACC)} curated genes:")
with pl.Config(tbl_rows=30, tbl_width_chars=180):
    print(summary)

ROWS = (
    CORE.select("hgnc_symbol", "mhc_class")
    .unique()
    .sort(["mhc_class", "hgnc_symbol"])["hgnc_symbol"]
    .to_list()
)
STATE_COLOR = {
    "not called": "#EDEDED",
    "called only": "#F2C57C",
    "boundary_matched": "#D65F5F",
}

fig, axes = plt.subplots(1, 2, figsize=(15.5, 8.4), sharey=True)
for ax, grp in zip(axes, ["comparison tools", "kmerseek HP"]):
    for j, sp in enumerate(SP):
        for i, sym in enumerate(ROWS):
            r = gene_state.filter(
                (pl.col("group") == grp)
                & (pl.col("species") == sp)
                & (pl.col("hgnc_symbol") == sym)
            )
            st = r["state"][0] if r.height else "not called"
            rr = reach_gene.filter(
                (pl.col("species") == sp) & (pl.col("hgnc_symbol") == sym)
            )
            reachable = bool(rr["n_reachable_families"][0] > 0) if rr.height else False
            # Background = whether the answer key has anything to score against here.
            ax.add_patch(
                plt.Rectangle(
                    (j - 0.5, i - 0.5),
                    1,
                    1,
                    facecolor="white" if reachable else "#BFC6CC",
                    edgecolor="white",
                    linewidth=1.2,
                )
            )
            if not reachable:
                ax.add_patch(
                    plt.Rectangle(
                        (j - 0.5, i - 0.5),
                        1,
                        1,
                        facecolor="none",
                        edgecolor="#8A9299",
                        linewidth=0.0,
                        hatch="///",
                    )
                )
            ax.add_patch(
                plt.Rectangle(
                    (j - 0.33, i - 0.33),
                    0.66,
                    0.66,
                    facecolor=STATE_COLOR[st],
                    edgecolor="#666666",
                    linewidth=0.4,
                )
            )
    ax.set_xlim(-0.5, len(SP) - 0.5)
    ax.set_ylim(len(ROWS) - 0.5, -0.5)
    ax.set_xticks(range(len(SP)))
    ax.set_xticklabels([mu.SPECIES_LABELS[s] for s in SP], fontsize=8)
    ax.set_title(
        f"{grp} (best over the group)" if grp != "kmerseek HP"
        else "kmerseek HP arms: hp_pbotc_1st_ed2 k=19/21/24, hp_thomas_dill2 k=26 (best over the four)",
        fontsize=10.5,
    )
    for s_ in ("right", "top", "left", "bottom"):
        ax.spines[s_].set_visible(False)
    ax.tick_params(length=0)
axes[0].set_yticks(range(len(ROWS)))
axes[0].set_yticklabels(ROWS, fontsize=8.5)
handles = [
    mpatches.Patch(facecolor=v, edgecolor="#666666", label=STATE_LABEL[k])
    for k, v in STATE_COLOR.items()
]
handles.append(
    mpatches.Patch(
        facecolor="#BFC6CC",
        hatch="///",
        edgecolor="#8A9299",
        label="no family of this gene is annotated in that proteome",
    )
)
axes[0].legend(
    handles=handles,
    fontsize=8,
    loc="upper left",
    bbox_to_anchor=(0, -0.06),
    ncol=2,
    framealpha=0.95,
)
sm = {(r["group"], r["species"]): r for r in summary.iter_rows(named=True)}
mu.finish_figure(
    fig,
    FIG_DIR / "224_detection_reachability_matrix.png",
    tools=mu.tools_text(mu.COMPARISON_TOOLS, kmerseek=mu.kmerseek_labels(HP_ARMS), note="each panel pooled over its group"),
    hypothesis=(
        "Where an MHC molecule can be boundary matched is decided by whether Pfam annotates "
        "its family in the target proteome, not by divergence; TAP1/TAP2 should stay matched "
        "everywhere."
    ),
    conclusion=(
        f"Comparison tools boundary match {sm[('comparison tools', 'mouse')]['boundary_matched']} of "
        f"{len(CORE_ACC)} genes in mouse, {sm[('comparison tools', 'fly')]['boundary_matched']} in fly "
        f"(one annotated C1-set protein) and {sm[('comparison tools', 'ciona')]['boundary_matched']} in "
        f"ciona; kmerseek HP {sm[('kmerseek HP', 'mouse')]['boundary_matched']} in mouse and "
        f"{sm[('kmerseek HP', 'ciona')]['boundary_matched']} in ciona. The grey edge is the annotation, not the tools."
    ),
    title="Detection state against reachability. The grey column edge is the answer key running out, not the tools.",
    footer_y=-0.10,
)
```

The two cliffs coincide, and fly proves the causal direction.

The comparison tools boundary match all 24 chr6 genes in mouse, chicken and zebrafish. In
ciona, worm, yeast, arabidopsis and E. coli they boundary match exactly 2, TAP1 and TAP2, and
every MHC molecule sits in the "called only" state. The searches do not stop: calls are still made on every gene at 2,000
Mya. What stops is the answer key. PF00129, PF00993 and PF00969 have zero annotated instances in
all six of those proteomes, and PF07654 has zero in five of them.

Fly is the exception and it is the most informative cell in the figure. At 600 Mya, further out
than ciona at 550, the comparison tools boundary match 23 of the 25 genes. The reason is a single fly protein.
Q9W257 (`BcDNA:RH44966`, 229 residues) is the only entry in the whole fly proteome annotated
with PF07654, and it carries no other Pfam family. That one protein turns 0 boundary-matched MHC
molecules into 20, in five of the nine tools. Remove it and fly looks like ciona.

Detection here is not a function of divergence time. It is a function of whether the answer key
contains one suitable protein, and notebook [223](./223_mhc_platform_vs_ig_domain.ipynb) §5–§6
says which kind: one whose aligned extent stops near the domain's end. Q9W257 is the
single-domain case, in a proteome that has exactly one of them.

So the benchmark cannot separate "MHC molecules stopped being findable" from "Pfam stopped
annotating MHC molecules". Any MHC recall number quoted past zebrafish is measuring the second
while sounding like the first, and the fly column shows the measurement moving with the
annotation rather than with the biology.

TAP1 and TAP2 are the control and they behave the other way. Their families PF00664 and PF00005
are annotated in all nine proteomes, and the comparison tools boundary match both genes in all
nine, E. coli included. Distance alone does not end detection at 550 Mya; the missing denominator does.

The ciona column deserves the extra sentence because it is where the two failure modes look
identical. Here the Pfam map is healthy (20,234 domain instances over 10,658 ciona proteins in
5,542 families, with the TAP control firing inside the same map) and it contains no
MHC platform and no C1-set. That is a real denominator fact. In the same run's Swiss-Prot truth
set, which this notebook does not use, ciona has almost no reviewed entries at all, and the
resulting curation gap produced a ciona cliff for every tool that looked exactly like this one.
An answer key can manufacture an evolutionary signal that is not there, which is the same
circularity as above with an independent instance.

TAPBP is worth a look for the same reason. It carries only PF07654, so it drops out with the
class II molecules at ciona even though it is a chaperone rather than an antigen receptor.

## 2. Called, and not boundary matched: the split runs opposite ways

Section 1 pooled tools. This splits every arm separately (nine comparison tools, seven kmerseek
arms), over the whole 221-gene MHC window rather than the curated molecules, into the three
states. The joint distribution is what is reported: not called, called but not boundary
matched, boundary matched, summing to 221 for every bar.

The claim being tested is that kmerseek finds the territory and misses the fenceline while the
profile methods do the reverse. Reporting only "kmerseek makes calls on 220 of 221 genes" or
only "kmerseek boundary matches 110 to phmmer's 204" would each support a different story, so
both go in the same bar.


```python
gl = mu.load_gene_level()
grid = (
    window.select(query_acc="accession")
    .join(ARMS, how="cross")
    .join(pl.DataFrame({"species": SP}), how="cross")
)
states = (
    grid.join(
        gl.select("query_acc", "tool", "variant", "species", "n_calls", "n_tp"),
        on=["query_acc", "tool", "variant", "species"],
        how="left",
    )
    .with_columns(
        n_calls=pl.col("n_calls").fill_null(0), n_tp=pl.col("n_tp").fill_null(0)
    )
    .with_columns(
        state=pl.when(pl.col("n_tp") > 0)
        .then(pl.lit("boundary_matched"))
        .when(pl.col("n_calls") > 0)
        .then(pl.lit("called only"))
        .otherwise(pl.lit("not called")),
        arm=mu.arm_label(pl.col("tool"), pl.col("variant")),
    )
)
tab = (
    states.group_by("species", "arm", "tool", "state")
    .len()
    .pivot(on="state", index=["species", "arm", "tool"], values="len")
    .fill_null(0)
)
for c in ["boundary_matched", "called only", "not called"]:
    if c not in tab.columns:
        tab = tab.with_columns(pl.lit(0).alias(c))
tab = tab.with_columns(
    total=pl.col("boundary_matched") + pl.col("called only") + pl.col("not called")
)
print(f"Every bar sums to the {window.height} MHC-window genes. Mouse:")
with pl.Config(tbl_rows=20, tbl_width_chars=180, fmt_str_lengths=42):
    print(
        tab.filter(pl.col("species") == "mouse")
        .select("arm", "not called", "called only", "boundary_matched", "total")
        .sort("boundary_matched", descending=True)
    )
print("\nBoundary matched given called, mouse: the fenceline question with the sensitivity")
print("difference divided out:")
cond = (
    tab.filter(pl.col("species") == "mouse")
    .with_columns(called=pl.col("called only") + pl.col("boundary_matched"))
    .with_columns(delim_given_called=(pl.col("boundary_matched") / pl.col("called")).round(3))
    .select("arm", "called", "boundary_matched", "delim_given_called")
    .sort("delim_given_called", descending=True)
)
with pl.Config(tbl_rows=20, tbl_width_chars=180, fmt_str_lengths=42):
    print(cond)

# Split the "called but not boundary matched" mass into two causes: the arm never called a family
# the gene actually has, or it called the right family and missed the boundary.
tpairs = (
    mu.window_truth()
    .select("query_acc", "pfam_id")
    .unique()
    .with_columns(right=pl.lit(True))
)
raw = (
    mu.focus_calls()
    .filter((pl.col("species") == "mouse") & pl.col("query_acc").is_in(WIN_ACC))
    .select("query_acc", "pfam_id", "tool", "variant", "is_tp")
    .collect()
    .join(tpairs, on=["query_acc", "pfam_id"], how="left")
    .with_columns(
        right=pl.col("right").fill_null(False),
        arm=mu.arm_label(pl.col("tool"), pl.col("variant")),
    )
)
cause = (
    raw.group_by("arm")
    .agg(
        genes_called=pl.col("query_acc").n_unique(),
        genes_right_family=pl.col("query_acc").filter(pl.col("right")).n_unique(),
        genes_boundary_matched=pl.col("query_acc").filter(pl.col("is_tp")).n_unique(),
    )
    .with_columns(
        delim_given_right_family=(
            pl.col("genes_boundary_matched") / pl.col("genes_right_family")
        ).round(3)
    )
    .sort("genes_boundary_matched", descending=True)
)
print(
    "\nmouse, from the raw calls: reaching the right family vs drawing the right box."
)
print(
    "The last column is the boundary question with family assignment divided out too:"
)
with pl.Config(tbl_rows=20, tbl_width_chars=190, fmt_str_lengths=42):
    print(cause)

ARM_ORDER = (
    tab.filter(pl.col("species") == "mouse")
    .sort("boundary_matched", descending=True)["arm"]
    .to_list()
)
fig, axes = plt.subplots(3, 3, figsize=(16.5, 12.5), sharex=True)
for ax, sp in zip(axes.ravel(), SP):
    t = tab.filter(pl.col("species") == sp)
    y = np.arange(len(ARM_ORDER))
    d = np.array(
        [t.filter(pl.col("arm") == a)["boundary_matched"][0] for a in ARM_ORDER], float
    )
    c = np.array(
        [t.filter(pl.col("arm") == a)["called only"][0] for a in ARM_ORDER], float
    )
    n = np.array(
        [t.filter(pl.col("arm") == a)["not called"][0] for a in ARM_ORDER], float
    )
    ax.barh(y, d, color="#D65F5F", label=STATE_LABEL["boundary_matched"])
    ax.barh(y, c, left=d, color="#F2C57C", label=STATE_LABEL["called only"])
    ax.barh(y, n, left=d + c, color="#EDEDED", label=STATE_LABEL["not called"])
    ax.set_yticks(y)
    ax.set_yticklabels(ARM_ORDER, fontsize=6.8)
    ax.invert_yaxis()
    ax.set_title(f"{sp}  ({mu.SPECIES_MYA[sp]} Mya)", fontsize=10)
    for s_ in ("right", "top"):
        ax.spines[s_].set_visible(False)
axes[2, 1].set_xlabel(f"MHC-window genes (of {window.height})")
axes[0, 0].legend(fontsize=8, loc="lower right", framealpha=0.95)
tm = {r["arm"]: r for r in tab.filter(pl.col("species") == "mouse").iter_rows(named=True)}
cm = {r["arm"]: r for r in cause.iter_rows(named=True)}
K19L = "kmerseek hp_pbotc_1st_ed2 k=19"
mu.finish_figure(
    fig,
    FIG_DIR / "224_called_vs_boundary_matched.png",
    tools=mu.tools_text(mu.COMPARISON_TOOLS, kmerseek=mu.kmerseek_labels(KM_ARMS)),
    hypothesis=(
        "kmerseek finds the territory and misses the fenceline: it calls on as many genes "
        "as the comparison tools but fails at the boundary step."
    ),
    conclusion=(
        f"Mouse: {K19L} calls on {tm[K19L]['called only'] + tm[K19L]['boundary_matched']} of "
        f"{window.height} genes (phmmer {tm['phmmer']['called only'] + tm['phmmer']['boundary_matched']}) "
        f"and boundary matches {tm[K19L]['boundary_matched']} (phmmer {tm['phmmer']['boundary_matched']}). Given a "
        f"call carrying a family the gene has, it boundary matches "
        f"{cm[K19L]['delim_given_right_family']:.0%} (phmmer {cm['phmmer']['delim_given_right_family']:.0%}). "
        "The loss is the family, not the fenceline."
    ),
    title="Not called / called but not boundary matched / boundary matched, per arm per species",
)
```

The stated claim is wrong in this run, and the correction matters more than the claim did.

kmerseek is not the arm that finds the territory and misses the fenceline. In mouse, kmerseek
`hp_pbotc_1st_ed2 k=19` makes calls on 220 of 221 window genes, as many as phmmer's 219, and
boundary matches 110 against phmmer's 204. Conditioned on having called a gene at all, phmmer
boundary matches 93% of them and kmerseek k19 50%. So kmerseek does lose on the joint outcome.

But the loss is not in the boundary. Notebook [223](./223_mhc_platform_vs_ig_domain.ipynb) §9
measures the boundary directly and finds it no worse than the comparison tools' once the
same domains are compared (§6 and 223 §9b). The split, counted from the raw calls on the same
genes, is that kmerseek k19 makes a call carrying a family the gene has on 117 of the 220
genes it touched, and boundary matches 110 of those 117, 94.0%, against phmmer's 204 of 218,
93.6%.
The same holds for the other kmerseek arms: 0.95, 0.95 and 0.91 for k21, k24 and thomas_dill
k26. What kmerseek misses is the family, not the fenceline.

So the Pareto statement is real but on a different axis than advertised. kmerseek is
lower-sensitivity in family assignment and no worse in boundary placement. The axis it loses is
the one that carries the headline recall number.

The other pattern in the panels goes the other way. The kmerseek arms hold their "called only"
mass out to E. coli while the comparison tools' collapses, which is 221's finding restated: whether
those distant calls mean anything is the subject of §5, and this figure makes no claim about
them.

## 3. The whole IoU curve, not a point on it

Section 2 used the run's threshold, IoU ≥ 0.5. This figure draws every arm: the nine comparison
tools in grey (Folddisco in green, because it is scored on coverage) and the seven kmerseek arms
in red. Notebook 223 showed that threshold interacts
badly with short domains inside multi-domain proteins, which raises the obvious question of
what happens at other thresholds.

The whole function is shown here, from 0.05 to 0.95, recomputed from the raw `iou` column
rather than from the fixed `is_tp` flag. Showing the function is legitimate; picking a
favourable point on it and quoting that instead of 0.5 would be metric-shopping, and nothing
below does it. The 0.5 line stays marked and every conclusion elsewhere in these notebooks
stays read off it.

The denominator is the 412 distinct (gene, family) truth pairs on the 221 MHC-window genes.
Distinct pairs rather than the 599 instances, because the run's `iou` column is already a
maximum over the instances of a family on a protein, so per-instance recall is not recoverable
from it.

```python
WT = mu.window_truth().select("query_acc", "pfam_id").unique()
print(
    f"denominator: {WT.height} distinct (gene, family) truth pairs on "
    f"{WT['query_acc'].n_unique()} MHC-window genes"
)

THS = [round(float(x), 2) for x in np.arange(0.05, 1.0, 0.05)]


def sweep(species):
    best = (
        mu.focus_calls()
        .filter(pl.col("query_acc").is_in(WIN_ACC) & (pl.col("species") == species))
        .group_by("tool", "variant", "query_acc", "pfam_id")
        .agg(best_iou=pl.col("iou").max())
        .collect()
    )
    g = (
        WT.join(ARMS, how="cross")
        .join(best, on=["tool", "variant", "query_acc", "pfam_id"], how="left")
        .with_columns(
            best_iou=pl.col("best_iou").fill_null(0.0),
            arm=mu.arm_label(pl.col("tool"), pl.col("variant")),
        )
    )
    rows = []
    for (arm,), s in g.group_by(["arm"]):
        v = s["best_iou"].to_numpy()
        rows += [
            {"arm": arm, "threshold": t, "recall": float((v >= t).mean())} for t in THS
        ]
    return pl.DataFrame(rows)


sw_mouse, sw_zf = sweep("mouse"), sweep("zebrafish")
SHOW = [
    "phmmer",
    "HHblits",
    "MMseqs2 iterative",
    "Foldseek",
    "Reseek",
    "Folddisco",
    "kmerseek hp_pbotc_1st_ed2 k=19",
    "kmerseek hp_thomas_dill2 k=26",
    "kmerseek protein20 k=10",
]
print("\nmouse, recall against IoU threshold:")
with pl.Config(tbl_rows=25, tbl_width_chars=210, fmt_str_lengths=30):
    print(
        sw_mouse.filter(pl.col("arm").is_in(SHOW))
        .with_columns(recall=pl.col("recall").round(3))
        .pivot(on="arm", index="threshold", values="recall")
    )

# Where, if anywhere, does a kmerseek curve cross a baseline curve?
KREF = "kmerseek hp_pbotc_1st_ed2 k=19"
cross_rows = []
for label, sw in [("mouse", sw_mouse), ("zebrafish", sw_zf)]:
    k = sw.filter(pl.col("arm") == KREF).sort("threshold")
    for arm in sw["arm"].unique().to_list():
        if arm.startswith("kmerseek"):
            continue
        b = sw.filter(pl.col("arm") == arm).sort("threshold")
        diff = k["recall"].to_numpy() - b["recall"].to_numpy()
        sign = np.sign(diff)
        idx = np.where(np.diff(sign) > 0)[0]
        if len(idx):
            i = int(idx[0])
            cross_rows.append(
                {
                    "species": label,
                    "comparison_tool": arm,
                    "crosses_between": f"{THS[i]}–{THS[i + 1]}",
                    "kmerseek_at_0.5": round(float(k["recall"][9]), 3),
                    "tool_at_0.5": round(float(b["recall"][9]), 3),
                    "kmerseek_at_0.95": round(float(k["recall"][-1]), 3),
                    "tool_at_0.95": round(float(b["recall"][-1]), 3),
                }
            )
CROSS = (
    pl.DataFrame(cross_rows)
    if cross_rows
    else pl.DataFrame({"species": [], "comparison_tool": []})
)
print(f"\nComparison tools whose curve {KREF} crosses from below:")
with pl.Config(tbl_rows=20, tbl_width_chars=190, fmt_str_lengths=30):
    print(CROSS)

fig, axes = plt.subplots(1, 2, figsize=(15, 5.6), sharey=True)
for ax, (label, sw) in zip(
    axes, [("mouse (100 Mya)", sw_mouse), ("zebrafish (430 Mya)", sw_zf)]
):
    for arm in sw["arm"].unique().to_list():
        s = sw.filter(pl.col("arm") == arm).sort("threshold")
        if arm.startswith("kmerseek"):
            style = dict(color="#D65F5F", lw=2.0, alpha=0.9)
        elif arm == "Folddisco":
            style = dict(color="#6ACC65", lw=2.0, alpha=0.95)
        else:
            style = dict(color="#999999", lw=1.2, alpha=0.75)
        ax.plot(s["threshold"], s["recall"], **style)
    ax.axvline(0.5, color="#333333", lw=1.4, ls="--")
    ax.text(0.51, 0.94, "the run's threshold", fontsize=8.5, color="#333333")
    ax.set_xlabel("IoU threshold")
    ax.set_title(label)
    for s_ in ("right", "top"):
        ax.spines[s_].set_visible(False)
axes[0].set_ylabel(f"fraction of the {WT.height} true (gene, family) pairs recovered")
axes[0].legend(
    handles=[
        mpatches.Patch(color="#D65F5F", label="kmerseek arms (7: four HP, dayhoff6, funcgroups8, protein20)"),
        mpatches.Patch(color="#6ACC65", label="Folddisco (coverage-scored tool)"),
        mpatches.Patch(color="#999999", label="the other eight comparison tools"),
    ],
    fontsize=8.5,
    loc="upper right",
)
k_m = sw_mouse.filter(pl.col("arm") == KREF).sort("threshold")
p_m = sw_mouse.filter(pl.col("arm") == "phmmer").sort("threshold")
mu.finish_figure(
    fig,
    FIG_DIR / "224_iou_threshold_sweep.png",
    tools=mu.tools_text(mu.COMPARISON_TOOLS, kmerseek=mu.kmerseek_labels(KM_ARMS)),
    hypothesis=(
        "kmerseek's calls are tight, so its recall curve is flatter across IoU thresholds "
        "and crosses the comparison tools' curves as the threshold rises."
    ),
    conclusion=(
        f"Mouse, {KREF}: recall {k_m['recall'][0]:.2f} at IoU 0.05, {k_m['recall'][9]:.2f} at 0.5, "
        f"{k_m['recall'][-1]:.3f} at 0.95; phmmer {p_m['recall'][0]:.2f}, {p_m['recall'][9]:.2f}, "
        f"{p_m['recall'][-1]:.3f}. kmerseek crosses only Folddisco"
        + (f" (between IoU {CROSS.filter(pl.col('species') == 'mouse')['crosses_between'][0]})" if CROSS.height else "")
        + "; it stays below the other eight at every threshold."
    ),
    title="Recall as a function of the IoU threshold, MHC window",
)
```

There is one crossover and it is with Folddisco.

In mouse, kmerseek `hp_pbotc_1st_ed2 k=19` is below Folddisco at loose thresholds (0.46 against
0.90 at IoU 0.05), crosses it between 0.20 and 0.25, and stays above it all the way out: 0.37 against
0.23 at the run's 0.5, and 0.046 against 0.015 at 0.95. In zebrafish it crosses the same curve
much later, between 0.90 and 0.95, ending at 0.017 against 0.010. Against the other eight
comparison tools it is below at every threshold in both species.

The crossover is real and its interpretation is narrow. Folddisco's curve falls steeply because
Folddisco is a motif matcher whose regions are not meant to mark a boundary at all, and the pipeline
scores it on coverage rather than IoU for exactly that reason (notebook
[223](./223_mhc_platform_vs_ig_domain.ipynb) §7). Reading the crossover as "kmerseek beats
Folddisco" would apply to Folddisco the metric its own pipeline arm does not use. What it does
show is that kmerseek's recall is the flattest curve among the arms whose calls are short: it
loses less than any other arm between IoU 0.05 and IoU 0.95, because most of the calls it makes
are already tight.

Every other number in these notebooks stays read at 0.5.

## 4. The alphabet ordering survives a metric that dislikes the tool

Sections 2 and 3 are competitive comparisons and kmerseek loses them. That makes the next
question worth asking on purpose: within kmerseek, does the HP encoding still beat the
alternatives when scored by a rule that is bad for kmerseek overall?

This is an internal control, not a competitive claim. Nothing here is measured against a
comparison tool; the figure is kmerseek only, seven arms. If the HP advantage were an artifact of choosing a metric that flatters it, a metric
that does not flatter it should erase the ordering.

```python
km = (
    states.filter(pl.col("tool") == "kmerseek")
    .group_by("species", "variant")
    .agg(
        boundary_matched=(pl.col("state") == "boundary_matched").sum(),
        called=(pl.col("state") != "not called").sum(),
    )
)
KM_LABEL = {v: mu.arm_pretty("kmerseek." + v) for v in KM_ARMS}
wide = (
    km.pivot(on="species", index="variant", values="boundary_matched")
    .select(["variant"] + SP)
    .sort("mouse", descending=True)
)
print(f"MHC-window genes boundary matched (of {window.height}), kmerseek arms only:")
with pl.Config(tbl_rows=20, tbl_width_chars=190, fmt_str_lengths=42):
    print(wide)
hp_min = wide.filter(pl.col("variant").is_in(HP_ARMS))["mouse"].min()
p20 = wide.filter(pl.col("variant") == "protein20_k10_lcTrue")["mouse"][0]
print(f"\nmouse: worst HP arm {hp_min}, protein20_k10 {p20}, ratio {hp_min / p20:.2f}x")
wins = []
for sp in SP:
    col = wide[sp].to_list()
    variants = wide["variant"].to_list()
    hp = [c for v, c in zip(variants, col) if v in HP_ARMS]
    other = [c for v, c in zip(variants, col) if v not in HP_ARMS]
    wins.append(
        {
            "species": sp,
            "best_HP": max(hp),
            "worst_HP": min(hp),
            "best_non_HP": max(other),
            "best_HP_wins": max(hp) > max(other),
            "every_HP_wins": min(hp) > max(other),
        }
    )
W = pl.DataFrame(wins)
print(
    "\nTwo readings of the ordering. `best_HP_wins` compares the best HP arm against the"
)
print(
    "best non-HP arm; `every_HP_wins` is the strict form, worst HP against best non-HP."
)
with pl.Config(tbl_rows=20, tbl_width_chars=170):
    print(W)
print(
    f"\nbest HP beats best non-HP in {W['best_HP_wins'].sum()} of {W.height} species; "
    f"the strict form holds in {W['every_HP_wins'].sum()}."
)

fig, axes = plt.subplots(1, 2, figsize=(15, 5.2))
x = [mu.SPECIES_MYA[s] for s in SP]
for v in wide["variant"].to_list():
    ys = wide.filter(pl.col("variant") == v).select(SP).row(0)
    hp = v in HP_ARMS
    axes[0].plot(
        x,
        ys,
        "o-",
        lw=2.2 if hp else 1.3,
        ms=7 if hp else 5,
        color="#D65F5F" if hp else "#4878CF",
        alpha=0.95 if hp else 0.8,
        label=KM_LABEL.get(v, v),
    )
axes[0].set_xscale("log")
axes[0].set_xticks(x)
axes[0].set_xticklabels([mu.SPECIES_LABELS[s] for s in SP], fontsize=7.5)
axes[0].set_ylabel(f"MHC-window genes boundary matched (of {window.height})")
axes[0].set_title("kmerseek only: red = HP alphabets, blue = dayhoff6, funcgroups8, protein20")
axes[0].legend(fontsize=7.5, ncol=2, framealpha=0.95)
gap = W.with_columns(
    best_gap=pl.col("best_HP") - pl.col("best_non_HP"),
    strict_gap=pl.col("worst_HP") - pl.col("best_non_HP"),
)
gx = np.arange(gap.height)
axes[1].bar(
    gx - 0.19,
    gap["best_gap"],
    0.36,
    color="#D65F5F",
    alpha=0.9,
    label="best HP − best non-HP",
)
axes[1].bar(
    gx + 0.19,
    gap["strict_gap"],
    0.36,
    color="#F2C57C",
    alpha=0.9,
    label="worst HP − best non-HP",
)
axes[1].axhline(0, color="#333333", lw=1.0)
axes[1].set_xticks(gx)
axes[1].set_xticklabels([mu.SPECIES_LABELS[s] for s in gap["species"]], fontsize=7.5)
axes[1].set_ylabel("difference in genes boundary matched")
axes[1].set_title("above zero means the HP arms win; the strict form is the pale bar")
axes[1].legend(fontsize=8)
for ax in axes:
    for s_ in ("right", "top"):
        ax.spines[s_].set_visible(False)
mu.finish_figure(
    fig,
    FIG_DIR / "224_hp_ordering_internal_control.png",
    tools=mu.tools_text(kmerseek=mu.kmerseek_labels(KM_ARMS) + " (kmerseek only, no comparison tool)"),
    hypothesis=(
        "Within kmerseek, the HP alphabets boundary match more MHC-window genes than dayhoff6, "
        "funcgroups8 and protein20, on the IoU rule that costs kmerseek the comparison overall."
    ),
    conclusion=(
        f"Mouse: worst HP arm {hp_min} genes, protein20 k=10 {p20} ({hp_min / p20:.2f}x). The best HP "
        f"arm beats the best non-HP arm in {W['best_HP_wins'].sum()} of {W.height} species; every HP "
        f"arm beats every non-HP arm in {W['every_HP_wins'].sum()}."
    ),
)
```

The ordering holds, on a metric that costs kmerseek the comparison against every baseline.

In mouse the four HP arms boundary match 115, 112, 112 and 110 of the 221 window genes
against `protein20 k=10`'s 67 — the weakest HP arm is 1.64 times the plain amino-acid encoding.
`dayhoff6 k=12` sits between at 89 and `funcgroups8 k=12` at 82, so the ordering is
HP > dayhoff6 > funcgroups8 > protein20, which is what compressing the alphabet toward
hydrophobic/polar predicts.

The best HP arm beats the best non-HP arm in eight of the nine species: 115 to 89 in mouse, 45
to 34 in chicken, 54 to 48 in zebrafish, 24 to 13 in ciona, 29 to 24 in fly, 45 to 23 in worm,
16 to 6 in yeast and 10 to 7 in arabidopsis. In E. coli every kmerseek arm is at 0 and the
question does not arise.

The strict form is weaker and is reported as such. Asking whether *every* HP arm beats every
non-HP arm holds in only four species — mouse, ciona, worm and arabidopsis. In chicken and
zebrafish the worst HP arm ties `dayhoff6 k=12`, and in fly and yeast a non-HP arm beats the
worst HP arm outright. The HP alphabets are not uniformly better than dayhoff6; the best of them
is, consistently.

This is worth more than it looks. Section 2's metric is the one on which kmerseek boundary matches
110 window genes to phmmer's 204. The HP advantage is measured on that same unflattering rule and
survives it, so it is not an artifact of picking a metric that suits the method.

It is still an internal comparison. It says the HP encoding is the right choice among kmerseek's
alphabets on the MHC region. It says nothing about kmerseek against anything else, and §2 and
§3 already answered that question the other way.

## 5. What kmerseek is calling past the annotation floor

From ciona outward there is no MHC molecule in the answer key, and yet the kmerseek arms keep
making calls on MHC-window genes. Section 1 said the benchmark cannot score those. This section
counts them anyway, as a census, for the four HP kmerseek arms in ciona, fly and worm.

The gray zone is the pipeline's own category for a call whose residues lie mostly outside any
Pfam annotation on that query protein. The reasoning in `evaluate_domain_calls.py` is that Pfam
is silent outside its annotations rather than negative, so counting such a call as a false
positive would assert that the annotation looked there and found nothing. Those calls are
excluded from the precision denominator and are neither true nor false positives.

Nothing below is a claim that these calls are right. They are unscored by construction.
This is a table of what a hypothesis-generating follow-up would have to look at, with every
count carrying its n.

```python
CENSUS_SP = ["ciona", "fly", "worm"]
calls = (
    mu.focus_calls()
    .filter(
        pl.col("species").is_in(CENSUS_SP)
        & (pl.col("tool") == "kmerseek")
        & pl.col("variant").is_in(HP_ARMS)
        & pl.col("query_acc").is_in(WIN_ACC)
    )
    .select("species", "variant", "query_acc", "pfam_id", "is_gray", "is_tp", "score")
    .collect()
)
totals = (
    calls.group_by("species", "variant")
    .agg(n_calls=pl.len(), n_gray=pl.col("is_gray").sum(), n_tp=pl.col("is_tp").sum())
    .with_columns(
        n_scoreable=pl.col("n_calls") - pl.col("n_gray"),
        frac_gray=(pl.col("n_gray") / pl.col("n_calls")).round(3),
    )
    .sort("species", "variant")
)
print("Gray-zone exclusion, kmerseek HP arms, MHC-window genes:")
with pl.Config(tbl_rows=20, tbl_width_chars=190, fmt_str_lengths=42):
    print(totals)
print(
    f"\nTotal calls {calls.height:_}; gray-zone excluded {calls['is_gray'].sum():_} "
    f"({calls['is_gray'].mean():.1%}); scoreable {calls.height - calls['is_gray'].sum():_}."
)

names = mu.load_pfam_names()
gray = calls.filter(pl.col("is_gray"))
per_fam = gray.group_by("species", "pfam_id").agg(
    n_calls=pl.len(),
    n_genes=pl.col("query_acc").n_unique(),
    n_arms=pl.col("variant").n_unique(),
)
recur = (
    per_fam.group_by("pfam_id")
    .agg(
        n_species=pl.col("species").n_unique(),
        n_calls_total=pl.col("n_calls").sum(),
        n_genes_total=pl.col("n_genes").sum(),
    )
    .join(names, on="pfam_id", how="left")
    .sort(["n_species", "n_calls_total"], descending=True)
)
print(
    f"\nDistinct Pfam families among the gray-zone calls: {per_fam['pfam_id'].n_unique()}"
)
print(
    f"Families appearing in all three species: "
    f"{recur.filter(pl.col('n_species') == 3).height}"
)
print(
    "\nTop families by pooled gray-zone call count (n_species = how many of ciona/fly/worm):"
)
with pl.Config(tbl_rows=20, tbl_width_chars=200, fmt_str_lengths=50):
    print(
        recur.head(15).select(
            "pfam_id",
            "pfam_name",
            "n_species",
            "n_calls_total",
            "n_genes_total",
            "pfam_desc",
        )
    )
print(
    "\nNo true/false-positive claim attaches to any row above. These calls sit outside"
)
print(
    "Pfam's annotated territory on the query, so the benchmark has no opinion on them."
)

TOP = recur.head(10)["pfam_id"].to_list()
fig, axes = plt.subplots(1, 2, figsize=(15.5, 5.4))
y = np.arange(len(TOP))
w = 0.26
for i, sp in enumerate(CENSUS_SP):
    vals = [
        per_fam.filter((pl.col("species") == sp) & (pl.col("pfam_id") == f))[
            "n_calls"
        ].sum()
        for f in TOP
    ]
    axes[0].barh(
        y + (i - 1) * w,
        vals,
        w,
        label=sp,
        color=["#D65F5F", "#4878CF", "#6ACC65"][i],
        alpha=0.9,
    )
lab = dict(names.select("pfam_id", "pfam_name").iter_rows())
axes[0].set_yticks(y)
axes[0].set_yticklabels([f"{f} {lab.get(f, '')}" for f in TOP], fontsize=8)
axes[0].invert_yaxis()
axes[0].set_xlabel("gray-zone calls (HP arms pooled)")
axes[0].set_title("what the kmerseek HP arms call where Pfam is silent")
axes[0].legend(fontsize=8.5)
t = totals.sort("species", "variant")
y2 = np.arange(t.height)
axes[1].barh(y2, t["n_scoreable"], color="#4878CF", label="scoreable")
axes[1].barh(
    y2, t["n_gray"], left=t["n_scoreable"], color="#BFC6CC", label="gray zone, excluded"
)
axes[1].set_yticks(y2)
axes[1].set_yticklabels(
    [f"{s} · {mu.arm_pretty('kmerseek.' + v)}" for s, v in zip(t["species"], t["variant"])],
    fontsize=7.5,
)
axes[1].invert_yaxis()
axes[1].set_xlabel("calls on MHC-window genes")
axes[1].set_title("how much of the output the benchmark declines to score")
axes[1].legend(fontsize=8.5)
for ax in axes:
    for s_ in ("right", "top"):
        ax.spines[s_].set_visible(False)
top3 = recur.head(3)
mu.finish_figure(
    fig,
    FIG_DIR / "224_gray_zone_census.png",
    tools=mu.tools_text(kmerseek=mu.kmerseek_labels(HP_ARMS) + " (kmerseek only)"),
    hypothesis=(
        "The calls kmerseek keeps making past the annotation floor land in Pfam-silent "
        "territory and recur across species as short compositionally biased modules, not as "
        "MHC families."
    ),
    conclusion=(
        f"{gray.height:_} of {calls.height:_} HP-arm calls on MHC-window genes in ciona, fly and "
        f"worm ({calls['is_gray'].mean():.1%}) are gray zone, over {per_fam['pfam_id'].n_unique()} "
        f"families; the top three are {', '.join(f'{n} ({c})' for n, c in zip(top3['pfam_name'], top3['n_calls_total']))}. "
        "No MHC family appears. Nothing here is scored."
    ),
)
```

The recurring families are short repeat modules, which is the least interesting explanation and
the one the counts support.

Across ciona, fly and worm the four HP arms make 9,142 calls on MHC-window genes, of which
3,182 (34.8%) fall in the gray zone and are excluded from scoring; 5,960 remain scoreable. The
gray-zone calls spread over 327 distinct Pfam families, and 41 of those appear in all three
species. The families carrying the most are zf-C2H2 (PF00096, 504 calls over 170 gene-species
cells), WD40 (PF00400, 310), Collagen (PF01391, 302), TSP_1 (PF00090, 107), EGF (PF00008, 85)
and Ldl_recept_a (PF00057, 83). The gray fraction ranges from 22.0% (`hp_pbotc_1st_ed2 k=24` in
worm) to 55.1% (`hp_thomas_dill2 k=26` in worm) across the twelve arm-species cells.

Every one of those is a short tandem module with strong compositional bias, which is what a
k-mer method matches on when there is no real homology left, and the recurrence across three
unrelated invertebrates is what a compositional artifact looks like rather than what a
conserved cryptic domain looks like. No MHC-related family appears anywhere in the list.

That is not proof these calls are wrong; the gray zone exists precisely because the benchmark
cannot say. It is a reason not to present them as an unrecognised signal without an
HP-composition-matched null to compare them against, which this run does not have.

## 6. Which overlap metric kmerseek actually wins, and on what

§2 left two things unexamined. It said kmerseek k19 misses the family rather than the
fenceline, without showing what the missed calls look like, and it cited
[223](./223_mhc_platform_vs_ig_domain.ipynb) §9 for a boundary advantage that turned out to
be a pooling artifact. This section opens both.

First the anatomy of the yellow "called, not boundary matched" bar, for kmerseek
hp_pbotc_1st_ed2 k=19 alone, and the same genes named rather
than counted. Then the metric question the bar raises: IoU 0.5 is one rule for scoring an
interval, and if kmerseek is drawing sensible boxes under a different rule, swapping the rule
should show it. Four rules are compared (reached at all, coverage, containment, IoU), first
unconditionally over all 412 truth pairs, then like-for-like on the pairs kmerseek reaches,
because those two comparisons answer different questions and only one of them is about
geometry.

```python
# §2 said kmerseek k19 misses the family rather than the fenceline. This opens that sentence
# up: of the genes in the "called, not boundary matched" state, how many failed on the boundary and
# how many never carried a true family at all, and what do the calls on the famous MHC
# molecules actually look like.
K19 = "hp_pbotc_1st_ed2_k19_lcTrue"
PFAM_NAME = dict(mu.load_pfam_names().select("pfam_id", "pfam_name").iter_rows())
SYMBOL = dict(window.select("accession", "hgnc_symbol").iter_rows())
WTRUTH = mu.window_truth()
TRUE_PAIRS = WTRUTH.select("query_acc", "pfam_id").unique().with_columns(right=pl.lit(True))
gl_k19 = mu.load_gene_level().filter(pl.col("variant") == K19)

rows, per_species = [], {}
for sp in SP:
    state = (
        pl.DataFrame({"query_acc": WIN_ACC})
        .join(gl_k19.filter(pl.col("species") == sp).select("query_acc", "n_calls", "n_tp"),
              on="query_acc", how="left")
        .with_columns(n_calls=pl.col("n_calls").fill_null(0), n_tp=pl.col("n_tp").fill_null(0))
    )
    called_only = state.filter((pl.col("n_tp") == 0) & (pl.col("n_calls") > 0))["query_acc"].to_list()
    c = (
        mu.focus_calls()
        .filter((pl.col("variant") == K19) & (pl.col("species") == sp)
                & pl.col("query_acc").is_in(called_only))
        .select("query_acc", "pfam_id", "qstart", "qend", "iou", "score")
        .collect()
        .join(TRUE_PAIRS, on=["query_acc", "pfam_id"], how="left")
        .with_columns(right=pl.col("right").fill_null(False),
                      call_len=pl.col("qend") - pl.col("qstart"))
    )
    near = c.filter("right").group_by("query_acc").agg(best_iou=pl.col("iou").max())
    wrong_only = [a for a in called_only if a not in set(near["query_acc"])]
    # For the wrong-family genes: where does the call sit, ignoring the label it carries?
    placed = (
        c.filter(pl.col("query_acc").is_in(wrong_only))
        .join(WTRUTH.select("query_acc", t_start="domain_start", t_end="domain_end"),
              on="query_acc", how="inner")
        .with_columns(ov=(pl.min_horizontal("qend", "t_end")
                          - pl.max_horizontal("qstart", "t_start")).clip(lower_bound=0))
        .with_columns(pos_iou=pl.col("ov") / (pl.max_horizontal("qend", "t_end")
                                              - pl.min_horizontal("qstart", "t_start")),
                      inside=pl.col("ov") / pl.col("call_len"))
        # Total sort key, so the representative call does not change between runs.
        .sort(["pos_iou", "qstart", "qend"], descending=[True, False, False])
        .group_by("query_acc").agg(pl.all().first())
    )
    per_species[sp] = dict(near=near, placed=placed, calls=c)
    rows.append(dict(
        species=sp, mya=mu.SPECIES_MYA[sp], boundary_matched=int((state["n_tp"] > 0).sum()),
        called_only=len(called_only), near_miss=near.height, wrong_family_only=len(wrong_only),
        median_pos_iou=round(float(placed["pos_iou"].median()), 3) if placed.height else None,
        frac_inside_a_true_domain=round(float((placed["inside"] >= 0.999).mean()), 3) if placed.height else None,
        median_call_len=float(placed["call_len"].median()) if placed.height else None))
split = pl.DataFrame(rows)
print("What the kmerseek hp_pbotc_1st_ed2 k=19 'called, not boundary matched' bar is made of, per species:")
with pl.Config(tbl_rows=20, tbl_width_chars=200):
    print(split)

near_mouse = (per_species["mouse"]["near"]
              .with_columns(gene=pl.col("query_acc").replace_strict(SYMBOL, default=None))
              .sort("best_iou", descending=True))
print(f"\nmouse: the {near_mouse.height} genes that are boundary near-misses, right family "
      f"and IoU under 0.5:")
with pl.Config(tbl_rows=20, tbl_width_chars=140):
    print(near_mouse.select("gene", "query_acc", "best_iou"))

wrong_fams = (
    pl.concat([per_species[s]["calls"].filter(~pl.col("right")).select("pfam_id") for s in SP])
    .group_by("pfam_id").len().sort("len", descending=True).head(10)
    .with_columns(family=pl.col("pfam_id").replace_strict(PFAM_NAME, default=pl.col("pfam_id"))))
print("\nFamilies most often transferred onto a called-but-not-boundary-matched gene, all species:")
with pl.Config(tbl_rows=20, tbl_width_chars=140):
    print(wrong_fams)

FAMOUS = ["HLA-A", "HLA-DRB1", "MICA", "TNF", "TAP1", "HSPA1A"]
acc_of = {s: a for a, s in SYMBOL.items() if s in FAMOUS}
famous_calls = (
    mu.focus_calls()
    .filter((pl.col("variant") == K19) & (pl.col("species") == "mouse")
            & pl.col("query_acc").is_in(list(acc_of.values())))
    .select("query_acc", "pfam_id", "qstart", "qend", "score").collect()
    .join(TRUE_PAIRS, on=["query_acc", "pfam_id"], how="left")
    .with_columns(right=pl.col("right").fill_null(False)))
famous_truth = WTRUTH.filter(pl.col("query_acc").is_in(list(acc_of.values())))
print(f"\nmouse: every k19 call on HLA-A, and not one of them carries MHC_I, C1-set or "
      f"MHC_I_C. Mouse has "
      f"{int(mu.target_family_counts(['PF00129'], ['mouse'])['n_target_proteins'][0])} "
      f"proteins annotated PF00129, so the answer was in the index:")
with pl.Config(tbl_rows=20, tbl_width_chars=140):
    print(famous_calls.filter(pl.col("query_acc") == acc_of["HLA-A"])
          .with_columns(family=pl.col("pfam_id").replace_strict(PFAM_NAME, default=pl.col("pfam_id")),
                        length=pl.col("qend") - pl.col("qstart"))
          .select("family", "qstart", "qend", "length", "score")
          .sort("score", descending=True))

fig = plt.figure(figsize=(13.5, 11.0))
gs = fig.add_gridspec(2, 2, height_ratios=[1, 1.9], hspace=0.30, wspace=0.22)

ax = fig.add_subplot(gs[0, 0])
y = np.arange(split.height)
ax.barh(y, split["wrong_family_only"], color="#C44E52", alpha=0.9,
        label="no call carried a true family")
ax.barh(y, split["near_miss"], left=split["wrong_family_only"], color="#DD8452", alpha=0.95,
        label="right family, IoU just under 0.5")
ax.set_yticks(y)
ax.set_yticklabels([f"{s}  ({mu.SPECIES_MYA[s]} Mya)" for s in split["species"]], fontsize=8.5)
ax.invert_yaxis()
ax.set_xlabel("genes in the 'called, not boundary matched' bar")
ax.set_title("kmerseek hp_pbotc_1st_ed2 k=19: the yellow bar is a family miss, not a boundary miss", fontsize=10)
ax.legend(fontsize=8, loc="lower left", framealpha=0.95)

ax = fig.add_subplot(gs[0, 1])
placed_mouse = per_species["mouse"]["placed"]
ax.hist(placed_mouse["pos_iou"].to_numpy(), bins=np.arange(0, 1.01, 0.05), color="#C44E52",
        alpha=0.85, label=f"wrong-family genes (n={placed_mouse.height}), best overlap\n"
                          f"with any true domain, family ignored")
ax.scatter(near_mouse["best_iou"].to_numpy(), np.full(near_mouse.height, 3.0), color="#DD8452",
           s=60, zorder=5, label=f"the {near_mouse.height} right-family near-misses")
ax.axvline(0.5, color="k", ls="--", lw=1)
ax.text(0.51, ax.get_ylim()[1] * 0.9, "IoU 0.5", fontsize=8)
ax.set_xlabel("IoU with the true domain")
ax.set_ylabel("genes")
ax.set_title("mouse: how far off the intervals are", fontsize=10)
ax.legend(fontsize=7.5, loc="upper right")

ax = fig.add_subplot(gs[1, :])
lanes, lane_labels = [], []
for i, gene in enumerate(FAMOUS):
    acc = acc_of[gene]
    length = int(famous_truth.filter(pl.col("query_acc") == acc)["protein_length"][0])
    yy = len(FAMOUS) - i
    ax.plot([0, length], [yy, yy], color="#BBBBBB", lw=1.2, zorder=1)
    for r in famous_truth.filter(pl.col("query_acc") == acc).iter_rows(named=True):
        ax.add_patch(mpatches.Rectangle(
            (r["domain_start"], yy - 0.19), r["domain_end"] - r["domain_start"], 0.38,
            facecolor="#4878CF", alpha=0.30, edgecolor="#4878CF", zorder=2))
        ax.text((r["domain_start"] + r["domain_end"]) / 2, yy + 0.26,
                PFAM_NAME.get(r["pfam_id"], r["pfam_id"]), ha="center", fontsize=7,
                color="#2A4A8A")
    for n, r in enumerate(famous_calls.filter(pl.col("query_acc") == acc).sort("qstart")
                          .iter_rows(named=True)):
        ax.add_patch(mpatches.Rectangle(
            (r["qstart"], yy - 0.09), max(r["qend"] - r["qstart"], 1), 0.18,
            facecolor="#55A868" if r["right"] else "#C44E52", edgecolor="none", alpha=0.9,
            zorder=3))
        dy, mid = 0.20 + 0.13 * (n % 4), (r["qstart"] + r["qend"]) / 2
        ax.plot([mid, mid], [yy - 0.10, yy - dy + 0.02], color="#C6A0A0", lw=0.5, zorder=2)
        ax.text(mid, yy - dy, PFAM_NAME.get(r["pfam_id"], r["pfam_id"]), ha="center",
                va="top", fontsize=5.4, color="#7A2A2A")
    lanes.append(yy)
    lane_labels.append(f"{gene}\n{acc}  ({length} aa)")
ax.set_yticks(lanes)
ax.set_yticklabels(lane_labels, fontsize=8)
ax.set_ylim(0.15, len(FAMOUS) + 0.8)
ax.set_xlabel("residue position in the human query")
ax.set_title("mouse, kmerseek hp_pbotc_1st_ed2 k19: every call on six famous MHC-region "
             "genes\nblue = true Pfam domain, red = the call and the family it transferred "
             "(none is the right family)", fontsize=10)
ax.spines["left"].set_visible(False)
for a_ in fig.axes:
    for s_ in ("right", "top"):
        a_.spines[s_].set_visible(False)
sp_m = split.filter(pl.col("species") == "mouse").to_dicts()[0]
mu.finish_figure(
    fig,
    FIG_DIR / "224_k19_called_not_boundary_matched.png",
    tools=mu.tools_text(kmerseek=[mu.BEST_KMERSEEK]) + " (kmerseek only)",
    hypothesis=(
        "The genes kmerseek calls on but does not boundary match are boundary near-misses: "
        "right family, box slightly off."
    ),
    conclusion=(
        f"Mouse: of {sp_m['called_only']} such genes, {sp_m['near_miss']} are right-family "
        f"near-misses and {sp_m['wrong_family_only']} never carried a true family. Their calls are "
        f"k-length windows ({sp_m['median_call_len']:.0f} residues), {sp_m['frac_inside_a_true_domain']:.0%} "
        "lying inside a true domain but labelled from an unrelated target. HLA-A gets 12 calls, none MHC."
    ),
    layout=False,
    footer_y=-0.02,
)
```

```python
# Does another overlap metric change the verdict? Four of them, over the same 412 truth pairs:
# reached at all, cover (fraction of the domain the call covers), containment (fraction of the
# call that lies in the domain) and the run's IoU. All four require the right family.
TRUTH_SPANS = WTRUTH.select("query_acc", "pfam_id", t_start="domain_start", t_end="domain_end")
WT_PAIRS = WTRUTH.select("query_acc", "pfam_id").unique()
mouse_calls = (
    mu.focus_calls()
    .filter((pl.col("species") == "mouse") & pl.col("query_acc").is_in(WIN_ACC))
    .select("query_acc", "pfam_id", "qstart", "qend", "iou", "cover", "is_tp", "tool", "variant")
    .collect())
scored = (
    mouse_calls.join(TRUTH_SPANS, on=["query_acc", "pfam_id"], how="inner")
    .with_columns(ov=(pl.min_horizontal("qend", "t_end")
                      - pl.max_horizontal("qstart", "t_start")).clip(lower_bound=0))
    .with_columns(contain=pl.col("ov") / (pl.col("qend") - pl.col("qstart")))
    # Total sort key, so the representative truth instance is stable across runs.
    .sort(["ov", "t_start", "t_end"], descending=[True, False, False])
    .group_by("query_acc", "pfam_id", "qstart", "qend", "tool", "variant").agg(pl.all().first())
    .group_by("tool", "variant", "query_acc", "pfam_id")
    .agg(iou=pl.col("iou").max(), cover=pl.col("cover").max(), contain=pl.col("contain").max(),
         reached=(pl.col("ov") > 0).any()))
metric_grid = (
    WT_PAIRS.join(ARMS, how="cross")
    .join(scored, on=["tool", "variant", "query_acc", "pfam_id"], how="left")
    .with_columns(iou=pl.col("iou").fill_null(0.0), cover=pl.col("cover").fill_null(0.0),
                  contain=pl.col("contain").fill_null(0.0),
                  reached=pl.col("reached").fill_null(False),
                  arm=mu.arm_label(pl.col("tool"), pl.col("variant"))))
board = (metric_grid.group_by("arm")
         .agg(reached=pl.col("reached").mean().round(3),
              cover50=(pl.col("cover") >= 0.5).mean().round(3),
              contain50=(pl.col("contain") >= 0.5).mean().round(3),
              iou50=(pl.col("iou") >= 0.5).mean().round(3))
         .sort("iou50", descending=True))
print(f"mouse, recall over the {WT_PAIRS.height} truth pairs. No overlap metric is "
      f"unconditional; each is capped by 'reached at all':")
with pl.Config(tbl_rows=20, tbl_width_chars=180, fmt_str_lengths=34):
    print(board)

# Conditioned: one best call per pair, restricted to the pairs k19 reaches.
offs = (mu.call_offsets("mouse", WIN_ACC)
        .with_columns(arm=mu.arm_label(pl.col("tool"), pl.col("variant")),
                      abs_start=pl.col("d_start").abs(), abs_end=pl.col("d_end").abs(),
                      tlen=pl.col("t_end") - pl.col("t_start")))
best_per_pair = (offs.sort(pl.col("abs_start") + pl.col("abs_end"),
                           "qstart", "qend", "t_start")
                 .group_by("arm", "query_acc", "pfam_id").agg(pl.all().first()))
K19_LABEL = "kmerseek hp_pbotc_1st_ed2 k=19"
k19_pairs = best_per_pair.filter(pl.col("arm") == K19_LABEL).select("query_acc", "pfam_id")
matched = (best_per_pair.join(k19_pairs, on=["query_acc", "pfam_id"], how="inner")
           .with_columns(within10=(pl.col("abs_start") <= 10) & (pl.col("abs_end") <= 10)))
print(f"\nLike-for-like on the {k19_pairs.height} pairs k19 reaches with a right-family call:")
with pl.Config(tbl_rows=20, tbl_width_chars=200, fmt_str_lengths=34):
    print(matched.group_by("arm").agg(
        n=pl.len(), med_abs_start=pl.col("abs_start").median(),
        med_abs_end=pl.col("abs_end").median(), within10=pl.col("within10").mean().round(3),
        med_iou=pl.col("iou").median().round(3), iou50=(pl.col("iou") >= 0.5).mean().round(3),
        frac_below_25=(pl.col("iou") < 0.25).mean().round(3),
        frac_above_90=(pl.col("iou") >= 0.9).mean().round(3)).sort("iou50", descending=True))

LEN_BINS = ["<30", "30-60", "60-120", ">=120"]
binned = matched.with_columns(
    bin=pl.when(pl.col("tlen") < 30).then(pl.lit("<30"))
    .when(pl.col("tlen") < 60).then(pl.lit("30-60"))
    .when(pl.col("tlen") < 120).then(pl.lit("60-120"))
    .otherwise(pl.lit(">=120")))
by_len = (binned.group_by("arm", "bin").agg(within10=pl.col("within10").mean().round(3))
          .pivot(on="bin", index="arm", values="within10")
          .select(["arm"] + LEN_BINS).sort("<30", descending=True))
bin_n = {r["bin"]: r["len"] for r in
         binned.filter(pl.col("arm") == K19_LABEL).group_by("bin").len().iter_rows(named=True)}
print(f"\nBoth edges within 10 residues, by true domain length (pairs per bin: {bin_n}):")
with pl.Config(tbl_rows=20, tbl_width_chars=180, fmt_str_lengths=34):
    print(by_len)

precision = (mouse_calls.with_columns(arm=mu.arm_label(pl.col("tool"), pl.col("variant")))
             .group_by("arm").agg(n_calls=pl.len(), precision=pl.col("is_tp").mean())
             .sort("precision", descending=True))
print("\nCall-level precision on the same 221 genes. Not rank-aware: it is driven by how")
print("many calls an arm emits, and Reseek emits three million on 221 genes:")
with pl.Config(tbl_rows=20, tbl_width_chars=140, fmt_str_lengths=34):
    print(precision.with_columns(precision=pl.col("precision").round(4)))

def arm_colour(a):
    if a.startswith("kmerseek"):
        return "#C44E52"
    return "#4878CF" if a in ("phmmer", "jackhmmer", "HHblits", "MMseqs2",
                              "MMseqs2 iterative") else "#55A868"

fig = plt.figure(figsize=(14, 10.5))
gs = fig.add_gridspec(2, 2, hspace=0.42, wspace=0.28)

ax = fig.add_subplot(gs[0, 0])
o = board.sort("iou50")
y, h = np.arange(o.height), 0.2
for k, (c, lab) in enumerate([("reached", "reached at all (right family, any overlap)"),
                              ("cover50", "coverage >= 0.5 (half the domain covered)"),
                              ("contain50", "containment >= 0.5 (half the call inside)"),
                              ("iou50", "IoU >= 0.5 (the run's metric)")]):
    ax.barh(y + (k - 1.5) * h, o[c], height=h, label=lab, alpha=0.9)
ax.set_yticks(y)
ax.set_yticklabels(o["arm"], fontsize=7.5)
ax.set_xlabel(f"fraction of the {WT_PAIRS.height} truth pairs")
ax.set_title("no overlap metric rescues kmerseek unconditionally:\nevery one is capped by "
             "the pairs it reaches at all", fontsize=10)
ax.legend(fontsize=7, loc="lower right")

ax = fig.add_subplot(gs[0, 1])
SHOW = [K19_LABEL, "phmmer", "HHblits", "MMseqs2 iterative", "Foldseek", "Reseek"]
PANEL = dict(zip(SHOW, ["#C44E52", "#4878CF", "#8172B2", "#64B5CD", "#55A868", "#CCB974"]))
x, w = np.arange(len(LEN_BINS)), 0.13
for k, a in enumerate(SHOW):
    v = [float(by_len.filter(pl.col("arm") == a)[b][0]) for b in LEN_BINS]
    ax.bar(x + (k - 2.5) * w, v, width=w, label=a, alpha=0.92, color=PANEL[a],
           edgecolor="white", hatch="//" if a == K19_LABEL else None)
ax.set_xticks(x)
ax.set_xticklabels([f"{b}\n(n={bin_n[b]})" for b in LEN_BINS])
ax.set_xlabel("true domain length, residues")
ax.set_ylabel("both edges within 10 residues")
ax.set_title(f"like-for-like on the {k19_pairs.height} pairs k19 reaches:\nkmerseek's edge "
             f"is short domains only", fontsize=10)
ax.legend(fontsize=7)

ax = fig.add_subplot(gs[1, 0])
for a, c in [(K19_LABEL, "#C44E52"), ("phmmer", "#4878CF"), ("HHblits", "#8172B2"),
             ("Foldseek", "#55A868")]:
    v = np.sort(matched.filter(pl.col("arm") == a)["iou"].to_numpy())
    ax.plot(v, np.arange(1, len(v) + 1) / len(v), lw=2, color=c, label=a)
ax.axvline(0.5, color="k", ls="--", lw=1)
ax.text(0.505, 0.05, "IoU 0.5", fontsize=8)
ax.set_xlabel("IoU with the true domain")
ax.set_ylabel("fraction of pairs at or below")
ax.set_title("the same pairs: kmerseek's IoU is compressed,\nfewest disasters, fewest "
             "bullseyes", fontsize=10)
ax.legend(fontsize=8, loc="upper left")

ax = fig.add_subplot(gs[1, 1])
o = precision.sort("precision")
ax.barh(np.arange(o.height), o["precision"], color=[arm_colour(a) for a in o["arm"]], alpha=0.9)
ax.set_yticks(np.arange(o.height))
ax.set_yticklabels(o["arm"], fontsize=7.5)
ax.set_xscale("log")
ax.set_xlabel("true positives / calls made (log scale)")
ax.set_title("call-level precision: every kmerseek arm beats every comparison tool\n(it emits far "
             "fewer calls per gene)", fontsize=10)
for i, (a, n, p) in enumerate(o.iter_rows()):
    ax.text(p * 1.15, i, f"{n:_} calls", va="center", fontsize=6.5, color="#555555")
ax.set_xlim(right=1.0)
for a_ in fig.axes:
    for s_ in ("right", "top"):
        a_.spines[s_].set_visible(False)
bd = {r["arm"]: r for r in board.iter_rows(named=True)}
mt = matched.group_by("arm").agg(iou50=(pl.col("iou") >= 0.5).mean())
mt = {r["arm"]: r["iou50"] for r in mt.iter_rows(named=True)}
mu.finish_figure(
    fig,
    FIG_DIR / "224_which_overlap_metrics.png",
    tools=mu.tools_text(mu.COMPARISON_TOOLS, kmerseek=mu.kmerseek_labels(KM_ARMS), note="top-left and bottom-right: all 16 arms; other panels: kmerseek k=19 vs five comparison tools"),
    hypothesis=(
        "Some overlap rule other than IoU (coverage, containment, reached at all) lets "
        "kmerseek's headline recall stop being a sensitivity number."
    ),
    conclusion=(
        f"No: {K19_LABEL} reaches {bd[K19_LABEL]['reached']:.2f} of the {WT_PAIRS.height} pairs at all "
        f"(phmmer {bd['phmmer']['reached']:.2f}), and every rule is capped by that. On the pairs it does "
        f"reach it clears IoU 0.5 on {mt[K19_LABEL]:.0%} (phmmer {mt['phmmer']:.0%}), and its edge in "
        "placing both ends within 10 residues is confined to domains under 30 residues."
    ),
    layout=False,
    footer_y=-0.02,
)
```

Unconditionally, no. Conditionally, two metrics and one domain-length regime.

Every overlap metric is capped by how many pairs the arm reaches with the right family, and
for k19 that is 0.456 of the 412 against the comparison tools' 0.98. Loosening IoU to coverage or to
containment moves it from 0.369 to at most 0.456 and it still finishes last among the
non-Folddisco arms. There is no rule under which the headline number stops being a
sensitivity number, which is §2's conclusion restated in the language of the metric.

Given that it reached a pair, two things go the other way.

kmerseek clears IoU 0.5 on more of its reached pairs than any other arm: 0.809 at k19 and
0.814 at k21, against phmmer 0.755, jackhmmer and HHblits 0.723, MMseqs2 iterative 0.686,
Foldseek 0.468. The distribution behind that is compressed: 3.2% of its reached pairs fall
below IoU 0.25 against phmmer's 10.1% and Foldseek's 22.6%, but only 12.2% clear 0.9 against
phmmer's 46.8%. It rarely misses badly and rarely lands exactly. Containment says the same
thing about over-claiming: half the call inside the domain on 0.888 of reached pairs, against
phmmer's 0.778 and Foldseek's 0.573.

Call-level precision is the other one. Every kmerseek arm beats every comparison tool:
protein20 k=10 31.4%, dayhoff6 20.5%, k19 3.7%, then MMseqs2 1.4%, phmmer 1.0%, HHblits 0.23%,
Reseek 0.01%. This is the weakest of the three claims. It is not rank-aware and it moves with
how many calls an arm emits, so it belongs next to the run's own precision-recall curves
rather than instead of them.

The localisation claim has to be stated narrowly, because the wide version does not survive
matching. Per call, kmerseek looks like the tightest arm in the run; per pair, on the 188
pairs it reaches, HHblits is at 2 and 2 residues, phmmer 3.5 and 4, and k19 13 and 9.5.
[223](./223_mhc_platform_vs_ig_domain.ipynb) §9 now carries both numbers. What is real is
confined to short domains: on domains under 30 residues k19 puts both edges within 10 residues
on 95.5% of pairs, ahead of phmmer's 87.0% and MMseqs2's 4.5%, and it collapses past 60: 12.2%
against phmmer's 59.8% in the 60 to 120 bin, 3.0% against 39.4% at 120 and above. Only 22 of the
188 pairs sit in the winning bin, so it is a small sample making a mechanistic point: a region
is built from matching k-mers, and nothing extends one across a domain much longer than k.

So the defensible sentence is not "kmerseek draws the tighter box". It is that kmerseek draws
a box that overshoots less and clears the threshold more often once it finds the family, and
that on domains close to k in length it is the most accurate arm in the run at saying where
the domain stops.

Detection tracks the annotation, not the divergence. The nine comparison tools boundary
match all 25 curated genes in mouse, chicken and zebrafish, and exactly 2, TAP1 and TAP2,
in ciona, worm, yeast, arabidopsis and E. coli. Fly breaks the monotonic reading: at 600 Mya, further out than
ciona, the comparison tools boundary match 23 of 25, entirely because one fly protein (Q9W257, 229 residues,
C1-set as its only Pfam family) is the single PF07654 entry in that proteome. One annotated
protein moves the score from 0 to 20 MHC molecules. TAP1 and TAP2, whose ABC families are
annotated in all nine proteomes, are boundary matched in all nine including E. coli. So the benchmark
cannot separate "MHC molecules stopped being findable" from "Pfam stopped annotating them", and
any MHC recall number past zebrafish measures Pfam's coverage while reading like a statement
about homology.

"kmerseek finds the territory and misses the fenceline" is not what this run shows. In
mouse, kmerseek hp_pbotc_1st_ed2 k=19 calls 220 of 221 window genes against phmmer's 219 and
boundary matches 110 against phmmer's 204. But conditioned on making a call carrying a family the
gene has, kmerseek boundary matches 110 of 117 (94.0%) and phmmer 204 of 218 (93.6%).
kmerseek's loss is in family assignment, not boundary placement; §6 and notebook
[223](./223_mhc_platform_vs_ig_domain.ipynb) §9b show its boundaries are no worse than the
comparison tools' on the same domains and tightest only on domains under 30 residues. The axis
it loses is sensitivity, which is the axis the headline number reports.

One crossover on the IoU sweep, and it is with the tool that is not scored on IoU. kmerseek
hp_pbotc_1st_ed2 k=19 crosses Folddisco between IoU 0.20 and 0.25 in mouse and stays above it to 0.95. Against
the other eight comparison tools it is below at every threshold in both mouse and zebrafish.
Folddisco is scored on coverage in its own pipeline arm, so the crossover is a statement about how flat
kmerseek's curve is, not a win. Nothing here relaxes the 0.5 threshold used everywhere else.

The HP alphabet ordering survives the unflattering metric. On the same rule that gives
kmerseek 110 window genes to phmmer's 204, the four HP arms boundary match 115/112/112/110 in
mouse against protein20 k=10's 67, with dayhoff6 at 89 and funcgroups8 at 82. The best HP arm beats the
best non-HP kmerseek arm in eight of the nine species, the exception being E. coli where every
arm is at 0. The strict form, every HP arm beating every non-HP arm, holds in only four
species; in fly and yeast a non-HP arm beats the weakest HP arm. So the claim is that the best
HP encoding is consistently the best kmerseek alphabet here, not that all HP arms dominate. It
is an internal control and says nothing about kmerseek against other tools.

Past the annotation floor there is a census and no claim. In ciona, fly and worm the HP arms
make 9,142 calls on MHC-window genes and 3,182 of them (34.8%) land in the gray zone, spread
over 327 families of which 41 recur in all three species. The most frequent are zf-C2H2 (504),
WD40 (310), Collagen (302), TSP_1 (107), EGF (85) and Ldl_recept_a (83): short tandem modules
with strong compositional bias, which is what a k-mer method finds when there is no homology
left. No MHC-related family appears. These calls are unscored by construction and are reported
as something to look at, not as a finding.
