```python
import polars as pl


def sum_until_first_false(df):
    return df.select(
        [
            pl.col(col)
            .cast(pl.UInt32)
            .cum_sum()
            .sub(pl.col(col).cast(pl.UInt32).cum_min().cast(pl.UInt32))
            .eq(pl.col(col).cast(pl.UInt32).cum_sum())
            .eq(False)
            .cast(pl.UInt32)
            .cum_sum()
            .max()
            .alias(col)
            for col in df.columns
        ]
    )


# Assuming your dataframe is called 'df'
result = sum_until_first_false(df.select(same_scop_cols))
result
```

### Write function for sensitive to first FP

```python
n_scop_cols = "n_family	n_superfamily	n_fold	n_class".split()
n_scop_cols
```

```python
def sensitivity_until_first_false_positive(same_scop_cols, n_scop_cols):
    return [
        (pl.col(same_col).cast(pl.Float64).arg_min() / (pl.col(n_col) - 1))
        .first()
        .fill_null(0)
        .alias(same_col.replace("same", "sensitivity"))
        for same_col, n_col in zip(same_scop_cols, n_scop_cols)
    ]


sensitivity = (
    multisearch_with_n_groups.sort(sourmash_col, descending=True)
    # .head(1000)
    .group_by("query_scop_id").agg(
        sensitivity_until_first_false_positive(same_scop_cols, n_scop_cols)
    )
).fill_nan(0)
sensitivity
```

```python
sensitivity.filter(
    (pl.col("sensitivity_family") > 0.9) & (pl.col("sensitivity_superfamily") > 0.5)
).sort("query_scop_id")
```

## Summary

Polars rewrite of the sensitivity-until-first-false-positive metric from
[011](./011-pandas-explore-compute-sensitivity-to-first-false-positive.ipynb). Method development.

Joining the multisearch results to query metadata gives 1,761,706 annotated rows, and the vectorised
implementation produces sensitivity for 15,163 queries against the pandas version's per-query
`groupby.apply`. The core trick is `arg_min` over the boolean same-family / same-superfamily /
same-fold / same-class columns to find the first false positive in one pass instead of walking each
query's hit list in Python.

Note the query count: 15,163 here against 15,177 in the metadata. The 14 missing queries have no
rows in the multisearch output at all, so they drop out of the join rather than scoring zero.
Whether an absent query counts as sensitivity 0 or as missing is a choice that changes every
downstream AUC, and later notebooks in this branch differ on it. That is the same
covered-versus-all-queries distinction that separates notebook 065's near-zero AUCs from notebook
067's high ones.

This implementation became `sensitivity_until_first_false_positive.py`.
