### Use regex to extract family information

https://regex101.com/r/gKBJAr/1

```python
pattern = r"(?P<family>(?P<superfamily>(?P<fold>(?P<class>[a-z])\.\d+)\.\d+)\.\d+)"
lineages_extracted = multisearch.query_scop_lineage.str.extractall(pattern)
lineages_extracted = lineages_extracted.droplevel(-1)
print(lineages_extracted.shape)
lineages_extracted.head()
```

## Use fixed SCOP lookup file from Foldseek paper

```python
scop_fixed = pd.read_csv(
    "https://raw.githubusercontent.com/steineggerlab/foldseek-analysis/refs/heads/main/scopbenchmark/data/scop_lookup.fix.tsv",
    sep="\t",
    header=None,
    names=["scop_id", "scop_class"],
    index_col=0,
).squeeze()
print(scop_fixed.shape)
scop_fixed.head()
```

## Summary

First look at the SCOPe40 protein benchmark output, at protein k=10 from the October 2024 pipeline
run. Exploratory, superseded by everything downstream.

The multisearch CSV holds 16,725 rows over 15,177 distinct query domains, of which 15,177 are
self-matches and 716 have a single intersecting hash. Removing self-matches and single-hash hits
leaves **832 real pairs**, which is the number that motivated everything after: at protein k=10
almost nothing survives.

The rest is scaffolding that became shared code: the regex that splits a SCOPe lineage into
family / superfamily / fold / class, and the switch to the FoldSeek authors' fixed SCOPe lookup
(11,211 domains) rather than parsing lineages out of the ASTRAL headers, which disagree for some
domains. Both live in `scop_constants.py` and `scop_utils_polars.py` now.

The metric histograms and pairplots here are of the original sourmash column set
(`prob_overlap`, `prob_overlap_adjusted`, `containment_adjusted`, `tf_idf_score` and friends). That
column set has since been replaced; `prob_overlap` in particular is shown in notebook 115 to be a
joint k-mer frequency score rather than a probability, and notebook 116 finds it and `query_tfidf`
to be the two weakest ranking metrics available.
